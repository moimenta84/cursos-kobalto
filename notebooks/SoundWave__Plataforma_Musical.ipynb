## Ejercicio 1: Formateando la cartelera

**Objetivo:** Dominar métodos de strings y formateo de texto

El departamento de marketing necesita generar fichas formateadas de las canciones para la cartelera de la app. Debes procesar los textos del catálogo usando métodos de strings.



### Tareas

- Genera el encabezado de la cartelera: título en mayúsculas centrado en 50 caracteres
- Crea una ficha formateada para cada canción con f-strings (título, artista, duración mm:ss)
- Extrae las iniciales de cada artista usando split()
- Genera un slug para URL a partir del título (minúsculas, espacios por guiones)
- Cuenta cuántas canciones contienen la letra 'e' en su título
- Crea un buscador: dada una búsqueda parcial, encuentra coincidencias (case-insensitive)


````python
# EJERCICIO 1: Formateando la cartelera
# Dataset disponible: titulos, artistas, duraciones_seg, generos

# 1. Encabezado de la cartelera
# "CARTELERA SOUNDWAVE" centrado en 50 caracteres con "=" de relleno
print("CARTELERA SOUNDWAVE".center(50, "="))


# 2. Ficha formateada de cada canción
# Formato: "  Blinding Lights — The Weeknd (3:21)"
print("\n=== Fichas ===")
for titulo, artista, duracion in zip(titulos, artistas, duraciones_seg):
    minutos = duracion // 60
    segundos = duracion % 60
    print(f"  {titulo} — {artista} ({minutos}:{segundos:02d})")


# 3. Iniciales de cada artista
# "The Weeknd" → "TW", "Queen" → "Q"
print("\n=== Iniciales ===")
for artista in artistas:
    iniciales = "".join(palabra[0].upper() for palabra in artista.split())
    print(f"{artista} → {iniciales}")


# 4. Slug para URL de cada canción
# "Blinding Lights" → "blinding-lights"
print("\n=== Slugs ===")
for titulo in titulos:
    slug = "-".join(titulo.lower().split())
    print(f"{titulo} → {slug}")


# 5. Canciones que contienen la letra "e" en el título
print("\n=== Títulos con 'e' ===")
for titulo in titulos:
    if "e" in titulo.lower():
        print(titulo)


# 6. Buscador: busca canciones que contengan "light" (sin importar mayúsculas)
busqueda = "light"
print(f"\n=== Resultados para '{busqueda}' ===")
resultados = [titulo for titulo in titulos if busqueda.lower() in titulo.lower()]
if resultados:
    for titulo in resultados:
        print(titulo)
else:
    print("Sin resultados")
````

## Ejercicio 2: Sistema de recomendaciones

**Objetivo:** Dominar if/elif/else, bucles for/while y break/continue

SoundWave necesita un sistema de recomendaciones que clasifique canciones por popularidad, filtre por género y genere informes automáticos del catálogo.



### Tareas

- Clasifica cada canción por popularidad: Viral (>5M), Hit (>2M), Popular (>1M), Emergente
- Filtra el catálogo: muestra solo canciones de un género dado (ej: 'Rock')
- Encuentra la primera canción con más de 3 millones de reproducciones (usa break)
- Genera un informe: cuenta canciones por género usando un bucle
- Crea un ranking: ordena las canciones por reproducciones (de mayor a menor) con enumerate
- Bonus: busca pares de canciones del mismo género usando bucles anidados


````python
# EJERCICIO 2: Sistema de recomendaciones
# Dataset disponible: titulos, artistas, generos, reproducciones

# 1. Clasificar canciones por popularidad
# Viral (>5M), Hit (>2M), Popular (>1M), Emergente
print("=== Clasificación de popularidad ===")
for titulo, repro in zip(titulos, reproducciones):
    if repro > 5_000_000:
        categoria = "Viral"
    elif repro > 2_000_000:
        categoria = "Hit"
    elif repro > 1_000_000:
        categoria = "Popular"
    else:
        categoria = "Emergente"
    print(f"{titulo} - {categoria}")


# 2. Filtrar por género
genero_filtro = "Rock"
print(f"\n=== Canciones de {genero_filtro} ===")
for titulo, genero in zip(titulos, generos):
    if genero == genero_filtro:
        print(titulo)


# 3. Primera canción con más de 3M de reproducciones (usa break)
print("\n=== Primera canción viral ===")
for titulo, repro in zip(titulos, reproducciones):
    if repro > 3_000_000:
        print(f"{titulo} ({repro:,})")
        break
else:
    print("Ninguna canción supera los 3M")


# 4. Informe: contar canciones por género
print("\n=== Canciones por género ===")
contador = {}
for genero in generos:
    contador[genero] = contador.get(genero, 0) + 1

for genero, cantidad in contador.items():
    print(f"{genero}: {cantidad}")


# 5. Ranking por reproducciones (mayor a menor)
print("\n=== Ranking SoundWave ===")
# Pista: crea una lista de índices y ordénala
indices = list(range(len(titulos)))
indices.sort(key=lambda i: reproducciones[i], reverse=True)
for pos, i in enumerate(indices, start=1):
    print(f"{pos}. {titulos[i]} - {reproducciones[i]:,}")


# 6. Bonus: pares de canciones del mismo género
print("\n=== Pares del mismo género ===")
for i in range(len(titulos)):
    for j in range(i + 1, len(titulos)):
        if generos[i] == generos[j]:
            print(f"{titulos[i]} <-> {titulos[j]} ({generos[i]})")
````

## Ejercicio 3: Herramientas del equipo

**Objetivo:** Crear funciones reutilizables con parámetros, valores por defecto y return

Tu equipo necesita herramientas reutilizables. Vas a crear funciones que el resto del equipo pueda usar sin conocer los detalles internos: calcular estadísticas, filtrar canciones y formatear informes.



### Tareas

- Crea formatear_duracion(segundos) que convierta segundos a formato 'mm:ss'
- Crea ficha_cancion(indice) que devuelva un string con toda la info de una canción
- Crea filtrar_por_genero(genero, minimo=0) con parámetro por defecto para reproducciones mínimas
- Crea estadisticas_catalogo() que devuelva múltiples valores con return (total, media, máximo)
- Crea generar_informe(*indices) que acepte un número variable de índices con *args
- Crea una función que use las anteriores para generar un informe completo


````python
# EJERCICIO 3: Herramientas del equipo
# Dataset disponible: titulos, artistas, duraciones_seg, generos, reproducciones

# 1. Función para formatear duración
# 201 → "3:21"
def formatear_duracion(segundos):
    """Convierte segundos a formato m:ss."""
    minutos, resto = divmod(segundos, 60)
    return f"{minutos}:{resto:02d}"


# Prueba:
print(formatear_duracion(201))   # "3:21"
print(formatear_duracion(482))   # "8:02"


# 2. Función ficha de canción
# Recibe un índice y devuelve un string formateado
def ficha_cancion(indice):
    """Devuelve la ficha de la canción en la posición indicada."""
    return (
        f"{titulos[indice]} — {artistas[indice]} "
        f"({formatear_duracion(duraciones_seg[indice])}) | "
        f"{generos[indice]} | {reproducciones[indice]:,} reproducciones"
    )


# Prueba:
print(ficha_cancion(0))


# 3. Filtrar por género con mínimo de reproducciones (por defecto 0)
def filtrar_por_genero(genero, minimo=0):
    """Devuelve los títulos del género indicado con al menos `minimo` reproducciones."""
    return [
        titulo
        for titulo, gen, repro in zip(titulos, generos, reproducciones)
        if gen == genero and repro >= minimo
    ]


# Prueba:
print(filtrar_por_genero("Rock"))
print(filtrar_por_genero("Pop", 3_000_000))


# 4. Estadísticas del catálogo (devuelve total, media, índice del máximo)
def estadisticas_catalogo():
    """Devuelve (total de reproducciones, media, índice de la más reproducida)."""
    total = sum(reproducciones)
    media = total / len(reproducciones)
    idx_max = max(range(len(reproducciones)), key=lambda i: reproducciones[i])
    return total, media, idx_max


# Prueba:
total, media, idx_max = estadisticas_catalogo()
print(f"Total: {total:,} | Media: {media:,.0f} | Más escuchada: {titulos[idx_max]}")


# 5. Informe con *args (acepta varios índices)
def generar_informe(*indices):
    """Imprime la ficha de cada índice recibido."""
    if not indices:
        print("No se ha indicado ninguna canción")
        return
    for indice in indices:
        if 0 <= indice < len(titulos):
            print(f"  - {ficha_cancion(indice)}")
        else:
            print(f"  - Índice {indice} fuera de rango")


# Prueba:
generar_informe(0, 2, 6)


# 6. Informe completo usando todas las funciones anteriores
print("\n" + " INFORME SOUNDWAVE ".center(60, "="))

print("\nCatálogo completo:")
generar_informe(*range(len(titulos)))

total, media, idx_max = estadisticas_catalogo()
print("\nEstadísticas:")
print(f"  Total de reproducciones: {total:,}")
print(f"  Media por canción: {media:,.0f}")
print(f"  Más escuchada: {ficha_cancion(idx_max)}")

print("\nPor género:")
for genero in sorted(set(generos)):
    canciones = filtrar_por_genero(genero)
    print(f"  {genero} ({len(canciones)}): {', '.join(canciones)}")

print("\nPop con más de 3M:")
print(f"  {', '.join(filtrar_por_genero('Pop', 3_000_000))}")
````

- Implementa una función que recomiende canciones basándose en gustos compartidos
- Usa comprehensions para transformar datos: filtrar, mapear y crear diccionarios
- Integra todo: genera un informe de usuario con su playlist, gustos y recomendaciones


````python
# EJERCICIO 4: Motor de playlists
# Dataset: titulos, artistas, generos, reproducciones, usuarios, playlists_usuarios
def titulos_de(indices):
    """Convierte una colección de índices en una lista de títulos ordenada."""
    return [titulos[i] for i in sorted(indices)]


# 1. Catálogo como lista de tuplas
# Cada tupla: (titulo, artista, genero, reproducciones)
catalogo = list(zip(titulos, artistas, generos, reproducciones))

# Muestra las 3 primeras canciones del catálogo
print("=== Primeras 3 canciones del catálogo ===")
for titulo, artista, genero, repro in catalogo[:3]:
    print(f"{titulo} - {artista} ({genero}) | {repro:,} reproducciones")


# 2. Diccionario artista → lista de canciones
canciones_por_artista = {}
for titulo, artista, _, _ in catalogo:
    canciones_por_artista.setdefault(artista, []).append(titulo)

# Muestra las canciones de cada artista
print("\n=== Canciones por artista ===")
for artista, canciones in canciones_por_artista.items():
    print(f"{artista}: {', '.join(canciones)}")


# 3. Sets: géneros únicos + operaciones entre gustos de usuarios
generos_unicos = set(generos)
print("\n=== Géneros únicos ===")
print(", ".join(sorted(generos_unicos)))

playlist_ana = set(playlists_usuarios["ana_95"])
playlist_carlos = set(playlists_usuarios["carlos_dev"])

en_comun = playlist_ana & playlist_carlos
solo_ana = playlist_ana - playlist_carlos
solo_carlos = playlist_carlos - playlist_ana

print("\n=== Ana vs Carlos ===")
print(f"En común: {titulos_de(en_comun) or 'ninguna'}")
print(f"Solo Ana: {titulos_de(solo_ana)}")
print(f"Solo Carlos: {titulos_de(solo_carlos)}")


# 4. Función de recomendación
# Recomienda a un usuario canciones que le gustan a otro y no tiene
def recomendar_canciones(usuario_destino, usuario_fuente):
    """Devuelve los títulos que tiene usuario_fuente y no usuario_destino,
    ordenados de más a menos reproducciones."""
    destino = set(playlists_usuarios[usuario_destino])
    fuente = set(playlists_usuarios[usuario_fuente])
    candidatas = fuente - destino
    ordenadas = sorted(candidatas, key=lambda i: reproducciones[i], reverse=True)
    return [titulos[i] for i in ordenadas]


# Prueba: ¿Qué le recomendaría Ana a Carlos?
print("\n=== Recomendaciones de Ana para Carlos ===")
for titulo in recomendar_canciones("carlos_dev", "ana_95"):
    print(f"- {titulo}")


# 5. Comprehensions
# Lista: títulos de canciones con más de 2M de reproducciones
exitos = [titulo for titulo, repro in zip(titulos, reproducciones) if repro > 2_000_000]

# Dict: {titulo: reproducciones} de todo el catálogo
reproducciones_por_titulo = {titulo: repro for titulo, repro in zip(titulos, reproducciones)}

# Set: géneros de canciones premium
# (es_premium no figura en el dataset de este ejercicio: se comprueba antes de usarlo)
if isinstance(es_premium, list):
    generos_premium = {genero for genero, premium in zip(generos, es_premium) if premium}
else:
    generos_premium = None

print("\n=== Comprehensions ===")
print(f"Más de 2M: {exitos}")
print(f"Reproducciones por título: {reproducciones_por_titulo}")
if generos_premium is not None:
    print(f"Géneros premium: {generos_premium}")
else:
    print("Géneros premium: no disponible (la plataforma no carga la lista es_premium en este ejercicio)")


# 6. Informe de usuario completo
# Para un usuario dado, muestra: nombre, plan, su playlist, géneros favoritos y recomendaciones
def informe_usuario(usuario):
    datos = usuarios[usuario]
    playlist = playlists_usuarios[usuario]

    print(f"\n=== Informe de {usuario} ===")
    print(f"Nombre: {datos['nombre']}")
    print(f"Plan: {datos['plan'].capitalize()}")

    if not playlist:
        print("Playlist vacía: sin géneros favoritos ni recomendaciones")
        return

    print("Playlist:")
    for i in playlist:
        print(f"  - {titulos[i]} ({artistas[i]})")

    # Géneros favoritos: los más repetidos en su playlist (puede haber empate)
    conteo_generos = {}
    for i in playlist:
        conteo_generos[generos[i]] = conteo_generos.get(generos[i], 0) + 1
    maximo = max(conteo_generos.values())
    favoritos = sorted(g for g, n in conteo_generos.items() if n == maximo)
    print(f"Géneros favoritos: {', '.join(favoritos)}")

    # Recomendaciones: canciones de otros usuarios, de sus géneros favoritos, que aún no tiene
    propias = set(playlist)
    candidatas = set()
    for otro, canciones in playlists_usuarios.items():
        if otro != usuario:
            candidatas |= set(canciones)
    candidatas -= propias

    recomendadas = sorted(
        (i for i in candidatas if generos[i] in favoritos),
        key=lambda i: reproducciones[i],
        reverse=True,
    )
    if recomendadas:
        print("Recomendaciones:")
        for i in recomendadas:
            print(f"  - {titulos[i]} ({generos[i]}, {reproducciones[i]:,})")
    else:
        print("Recomendaciones: ninguna nueva en sus géneros favoritos")


informe_usuario("ana_95")
````
